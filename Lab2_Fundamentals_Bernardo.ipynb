```python
# [EXERCISE 1] SENSOR READING VALIDATION & CLASSIFICATION
ENGINEER = "Bernardo"
STUDENT_ID = "TUPM-26-4349"
SEED_NUM = int(STUDENT_ID[-1]) # 9
LIMIT = (SEED_NUM * 10) + 20 # 110

raw_inputs = ["10.5", "abc", str(LIMIT + 15), "3.2", "75.0", "xyz"]
valid_readings = []
invalid_count = 0
classification_results = []
execution_log = ["Initialization: Sensor validation pipeline active."]

for idx, item in enumerate(raw_inputs):
    try:
        val = float(item)
        valid_readings.append(val)
        execution_log.append(f"Reading {idx+1}: '{item}' parsed successfully to float.")
        
        if val > LIMIT:
            classification_results.append(f"{val} -> ALERT: Over Limit")
        elif val < 5:
            classification_results.append(f"{val} -> ALERT: Low Signal")
        else:
            classification_results.append(f"{val} -> STATUS: Nominal")
    except ValueError:
        invalid_count += 1
        execution_log.append(f"Reading {idx+1}: ERROR - '{item}' contains invalid non-numeric data.")

print("Assessment Data:")
print(f"Generated Sensor Data: {raw_inputs}")
print(f"Valid/Invalid Results: {len(valid_readings)} Valid, {invalid_count} Invalid")
print("Classification Results:")
for res in classification_results:
    print(f" {res}")
print("\nExecution Log:")
for log in execution_log:
    print(f" {log}")
print(f"\nFinal Output:\nPipeline Finished. Total Processed: {len(raw_inputs)}")
```

```python
# [EXERCISE 2] SIGNAL CHARACTER-BY-CHARACTER ANALYSIS
raw_signal = f" {ENGINEER}-{SEED_NUM}x{STUDENT_ID[-4:]}!! "
processed_signal = raw_signal.strip().upper()

char_analysis = []
for idx, char in enumerate(processed_signal):
    if char.isalpha():
        ctype = "Letter"
    elif char.isdigit():
        ctype = "Digit"
    elif char.isspace():
        ctype = "Space"
    else:
        ctype = "Special Character"
    char_analysis.append(f"Pos {idx}: '{char}' ({ctype})")

digit_count = sum(c.isdigit() for c in processed_signal)
alpha_count = sum(c.isalpha() for c in processed_signal)
signal_class = "Alphanumeric Telemetry" if (digit_count > 0 and alpha_count > 0) else "Pure Data Stream"

print("Assessment Data:")
print(f" Generated Signal : '{raw_signal}'")
print(f" Processed Signal : '{processed_signal}'")
print(" Character Analysis:")
for analysis in char_analysis[:8]: 
    print(f" {analysis}")
print(f" Signal Classification: {signal_class}")
print(f"\n Execution Log:\n [LOG] Raw Length: {len(raw_signal)} -> Processed Length: {len(processed_signal)}")
print(f"\n Final Output:\n Diagnostic Report for {processed_signal} successfully generated.")
```

```python
# [EXERCISE 3] ELECTRONIC LOCK PASSWORD & ATTEMPT LIMITER
ENGINEER = "Bernardo"
STUDENT_ID = "TUPM-26-4349"
SEED_NUM = int(STUDENT_ID[-1]) # Evaluates to 9

generated_password = f"{ENGINEER.upper()}{SEED_NUM}"
attempt_limit = (SEED_NUM % 3) + 3
simulated_attempts = ["BERNARDO", "ADMIN", f"{ENGINEER.upper()}{SEED_NUM}"]
attempts_made = 0
access_result = "DENIED"
final_system_state = "LOCKED"
execution_log_3 = []

for attempt in simulated_attempts:
    attempts_made += 1
    execution_log_3.append(f"Attempt {attempts_made}: Received input code '{attempt}'")
    
    if attempt == generated_password:
        access_result = "GRANTED"
        final_system_state = "UNLOCKED"
        execution_log_3.append("Success: Password matches master credential.")
        break
    else:
        execution_log_3.append("Failure: Invalid password string.")
        if attempts_made >= attempt_limit:
            final_system_state = "TERMINATED / LOCKOUT ACTIVE"
            execution_log_3.append("Security Alert: Max attempts reached.")
            break

print("Assessment Data:")
print(f"Generated Password: {generated_password}")
print(f"Attempt Limit : {attempt_limit}")
print(f"Attempts Made : {attempts_made}")
print(f"Access Result : {access_result}")
print(f"Final System State: {final_system_state}")
print("\nExecution Log:")
for log in execution_log_3:
    print(f" [SECURITY] {log}")
print(f"\nFinal Output:\nSystem status resolution: {final_system_state}")
```
